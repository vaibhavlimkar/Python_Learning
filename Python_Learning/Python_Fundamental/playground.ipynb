```python
# fibonnaci series
n = 5
a = 0
b = 1

print(a)
print(b)

for i in range(n-2):
    c = a+b
    print(c)

    a = b
    b = c
```

```python
n = int(input("How many Fibonacci numbers? "))

a, b = 0, 1

for _ in range(n):
    print(a)
    a, b = b, a + b
```

```python
n = 5
a,b = 0,1

count = 0

while count < n:
    print(a)
    a,b = b,a+b
    count = count + 1
```
