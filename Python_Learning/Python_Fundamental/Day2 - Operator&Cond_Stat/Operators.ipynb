#### Operator is some symbol or keyword that is used to perform some task

- Arithmetic Operators - +,-,*,/,//,%,
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators




```python
# Arithmetic Operator
# Addition
print(5+6) 
# Subtraction
print(3-2)
# Multiplication
print(5*3)
# Modulus
print(4%2) # Reminder - 0
# Floor
print(5/2) # Quotitent - 2.5
# Floor Division
print(5//2) # Quotitent - 2
```
