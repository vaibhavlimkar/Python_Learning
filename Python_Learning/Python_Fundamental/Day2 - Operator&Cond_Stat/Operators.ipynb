# Operator is some symbol or keyword that is used to perform some task

- Arithmetic Operators - +,-,*,/,//,%,
- Relational Operators - >,<,<=,>=,!=,==
- Logical Operators - AND, OR, NOT
- Bitwise Operators - Bitwise OR, Bitwise AND, Bitwise NOR, Btwise XOR, Bitwise NOT, Bitwise Left shift and Right Shift
- Assignment Operators - +=, -=, *=, %=, /=, //=
- Membership Operators - is, is not
- Identity Operators - in, not in 




```python
# Arithmetic Operator

print(4+5)

# Subtraction
print(5-2)

# Multiplication
print(5*2)

# Modulus
print(5%2) # Reminder - 1

# Floor 
print(5/2) # Quotient - 2.5

# Floor Division
print(5//2) # Quotient - 2

```

```python
# 2 - Relational Operators
# <,>,<=,>=,==,!=

print(3>5)
print(3<4)
print(4>=4)
print(5<=4)
print(5==5)
print(5!=3)
```

```python
# 3 - logical Operator
print(1 and 0)
print(1 or 0)
print(not 1)
print(4 and 5) # Both should be true
print(4 and 3 and 5) # Both should be true and last value will be as ouput - 5
print(2 or 3 or 5) # Atleast one value should be true here we have output - 2
```

```python
# 4 - Assignment Operator

x = 5
x+= 3 # x = x + 3
print(x)

y = 5
y-= 3 # x = x - 3
print(y)


```

```python
# 5 - Membership Operator
print('D' in 'Delhi')
print('D' not in 'Delhi')

print(1 in [1,2,4,54,"ABC"])
```

```python
# Membership Operator
a = [1,2,3]
b = [1,2,3]
c = a
print(a is c) # is checks whether the two names refer to the same object in memory or not
print(a is b) # If two list having same content it will give False because it is two different list
print(a == b) # == it checks whether two object content same variable or not 
print(a == c)
print(id(a)) # same address 259980843
print(id(b)) # same address 259980843
```
