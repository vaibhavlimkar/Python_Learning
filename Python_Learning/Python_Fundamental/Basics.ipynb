```python
# Python has Dynamic typing and Dynamic Binding

# Dynamic Typing - we don't decalre data type in python
a = 5
print(a)  
```

```python
# Dynamic Binding - Here one variable can store multiple values and we can use same multiple times

a = 1,'3',[1,2,3,4,5]
print(a)

a = {1:3,4:5.6} # If we have two varaible with same name so last variable has values will return those values
print(a)
```

```python
# Python Output
# end = '' - it is used when we combine two line into one line and '' is used add space or some other special character
print('Vaibhav','Limkar','Omee', end =' ')
print('Vaibhav','Limkar','Omee')
```

```python
# sep - If we want some specific thing in between two/multiple character we can use sep = '-' like this
print('Vaibhav','Omee',sep = '/')
```

```python
# Combination of end and sep
print('Vaibhav','Limkar',end = ' ',sep = '-')
print('Data Engineer','1000000 CR', sep ='-')
```
